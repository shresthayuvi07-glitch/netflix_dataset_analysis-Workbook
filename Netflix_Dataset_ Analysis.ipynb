### Q85. Explain in a markdown comment: why is **df.loc[condition, "col"] = value** the SAFE way to update, instead of **df[condition]["col"] = value**? (SettingWithCopyWarning!)

**Answer**: 

**df.loc[condition,"col"]=value** is the safe way to update because, when we use **.loc** the data will changed or update permanently and we don't need to use **inplace =True** again and again. However, if we just use, **df[condition, "col"]=value**, we have to use **inplace= True** eveytime, because without **.loc or inplace = True**, the data only changed temporarly.
